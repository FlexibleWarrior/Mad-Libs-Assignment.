```python
def mad_libs():
    print("\nWelcome to Mad Libs!")
    print("Choose a story template:")
    print("1. Fairy Tale")
    print("2. Space Adventure")
    print("3. Sports Story")
    
    choice = input("Enter your choice (1-3): ")
    
    print("\nPlease provide the following words:")
    
    if choice == "1":
        # Fairy Tale Story
        adjective = input("Enter an adjective: ")
        noun1 = input("Enter a noun: ")
        verb = input("Enter a verb (past tense): ")
        adverb = input("Enter an adverb: ")
        animal = input("Enter an animal: ")
        noun2 = input("Enter another noun: ")
        
        story = f"""
        Once upon a time, there was a {adjective} {noun1} who {verb} {adverb} through the enchanted forest. 
        They encountered a talking {animal} who granted them a magical {noun2}. 
        And so, our hero lived happily ever after.
        """
        
    elif choice == "2":
        # Space Adventure
        planet = input("Enter a planet name: ")
        alien = input("Enter an alien species: ")
        spaceship = input("Enter a spaceship name: ")
        verb = input("Enter a verb: ")
        gadget = input("Enter a futuristic gadget: ")
        
        story = f"""
        Captain of the {spaceship}, we've arrived at planet {planet}. 
        The {alien} are approaching! Quick, {verb} the {gadget} before it's too late!
        """
        
    elif choice == "3":
        # Sports Story
        sport = input("Enter a sport: ")
        athlete = input("Enter an athlete's name: ")
        number = input("Enter a number: ")
        adjective = input("Enter an adjective: ")
        body_part = input("Enter a body part: ")
        
        story = f"""
        In the final seconds of the {sport} championship, {athlete} scored {number} points!
        What a {adjective} performance! The crowd went wild, cheering at the top of their {body_part}s.
        """
        
    else:
        print("Invalid choice!")
        return
    
    print("\nHere's your Mad Lib story:")
    print(story)
    print("\nThanks for playing!")

# Run the game
mad_libs()
```
